# Operators & Conditional Statements
Performance Category Program:
1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).

```python
Score=int(input("Enter your score(0 to 10): "))
```

2. Find the performance category based on the input score using following criteria:
a. Above Average:  Score greater than 7
b. Average: Score between 4 and 7(both inclusive)
c. Below Average: Score lesser than 4

```python
if Score > 7:
  print("Above Average: Excellent performance! Keep it up")
elif Score >= 4:
  print("Average: Good effort! Keep practicing,there's room for improvement")
else:
  print("Below Average: Don't give up! keep practicing and focus on your weak areas")
```

3. Output: Print the Performance category

```python
Score=int(input("Enter your score(0 to 10): "))
if Score > 7:
  print("Above Average: Excellent performance! Keep it up")
elif Score >= 4:
  print("Average: Good effort! Keep practicing,there's room for improvement")
else:
  print("Below Average: Don't give up! keep practicing and focus on your weak areas")
```

# Explanation
The program takes a score and if statement check it is greater than 7 then prints Above average.If the score is not greater than 7 then it goes to elif condition the score is from 4 to 7 prints Average.If the score is below 4 it goes to else condition and prints Below Average
